# 墨尔本房价预测

## 将建立决策树和随机森林模型预测墨尔本市的房价。

## 第一步：决策树
1. 导入`pandas`库
2. 读取墨尔本市数据集
3. 预览数据集描述性统计摘要

```python
import pandas as pd
```

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("dansbecker/melbourne-housing-snapshot")

print("Path to dataset files:", path)
```

```python
melbourne_file_path = '/kaggle/input/melbourne-housing-snapshot/melb_data.csv'

melbourne_data = pd.read_csv(melbourne_file_path)

melbourne_data.describe()
```

4. 预览数据集列名

```python
melbourne_data.columns
```

5. 删除有空值的行

```python
melbourne_data = melbourne_data.dropna(axis=0)
```

6. 定义模型因变量`y`=价格`Price`

```python
y = melbourne_data.Price
```

7. 把影响价格的主要因素列赋值给`melbourne_features`
8. 定义模型自变量`X`=`melbourne_features`

```python
melbourne_features = ['Rooms', 'Bathroom', 'Landsize', 'Lattitude', 'Longtitude']
X = melbourne_data[melbourne_features]
```

9. 生成`X`变量述性统计摘要

```python
X.describe()
```

10. 展示`X`变量前五列

```python
X.head()
```

11. 导入决策树库
12. 定义模型。为random_state指定一个数字，以确保每次运行都得到相同的结果
13. 拟合模型

```python
from sklearn.tree import DecisionTreeRegressor

# 定义模型。为random_state指定一个数字，以确保每次运行都得到相同的结果
melbourne_model = DecisionTreeRegressor(random_state=1)

# 拟合模型
melbourne_model.fit(X, y)
```

14. 展示`X`变量前五列
15. 展示用`melbourne_model`模型预测`X`前五行变量的房价`y`变量`Price`

```python
print("Making predictions for the following 5 houses:")
print(X.head())
print("The predictions are")
print(melbourne_model.predict(X.head()))
```

### 用平均绝对误差`MAE`验证决策树模型精度

16. 导入`sklearn.metrics`库`mean_absolute_error`模块
17. 验证预测`X`前五行变量的房价`y`变量`Price`的和原始数据集前五行房价对比的平均绝对误差

```python
from sklearn.metrics import mean_absolute_error

predicted_home_prices = melbourne_model.predict(X)
mean_absolute_error(y, predicted_home_prices)
```

数据解读：由结果可以看出用数据集前五行训练出的预测房价与数据集前五行真实房价差值是1115.7467183128902元

18. 数据分割：

使用 train_test_split 将数据集 (X, y) 划分为训练集和验证集

random_state=0 确保每次运行都能得到相同的随机分割结果（可复现性）

19. 模型定义与训练：

创建决策树回归模型 (DecisionTreeRegressor)

使用训练数据 (train_X, train_y) 拟合模型

20. 模型验证：

用训练好的模型对验证集 (val_X) 进行预测

计算预测值 (val_predictions) 与真实值 (val_y) 的平均绝对误差 (MAE)

打印出这个平均绝对误差`MAE`作为模型性能评估

```python
from sklearn.model_selection import train_test_split

# split data into training and validation data, for both features and target
# The split is based on a random number generator. Supplying a numeric value to
# the random_state argument guarantees we get the same split every time we
# run this script.
train_X, val_X, train_y, val_y = train_test_split(X, y, random_state = 0)
# Define model
melbourne_model = DecisionTreeRegressor()
# Fit model
melbourne_model.fit(train_X, train_y)

# get predicted prices on validation data
val_predictions = melbourne_model.predict(val_X)
print(mean_absolute_error(val_y, val_predictions))
```

21. 创建函数，通过调用函数来帮助比较`max_leaf_nodes`不同值的`MAE`分数

```python
from sklearn.metrics import mean_absolute_error
from sklearn.tree import DecisionTreeRegressor

def get_mae(max_leaf_nodes, train_X, val_X, train_y, val_y):
    model = DecisionTreeRegressor(max_leaf_nodes=max_leaf_nodes, random_state=0)
    model.fit(train_X, train_y)
    preds_val = model.predict(val_X)
    mae = mean_absolute_error(val_y, preds_val)
    return(mae)
```

22. 添加实例`[5, 50, 500, 5000]`查看不同树的深度输出的`MAE`有什么不同，并选择最优数目深度

```python
# compare MAE with differing values of max_leaf_nodes
for max_leaf_nodes in [5, 50, 500, 5000]:
    my_mae = get_mae(max_leaf_nodes, train_X, val_X, train_y, val_y)
    print("Max leaf nodes: %d  \t\t Mean Absolute Error:  %d" %(max_leaf_nodes, my_mae))
```

数据解读：在列出的选项中，`500`的深度是最佳叶数。

## 第二步：随机森林

1. 用随机森林模型训练`train_test_split`将数据集

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error

forest_model = RandomForestRegressor(random_state=1)
forest_model.fit(train_X, train_y)
melb_preds = forest_model.predict(val_X)
print(mean_absolute_error(val_y, melb_preds))
```

数据解读，输出`MAE`结果是`207190.6873773146`，相比起决策树最优深度`500`的`MAE`结果`261718`有显著改善。

2. 使用随机森林模型训练`melb_data.csv`墨尔本房价完整数据集

```python
# To improve accuracy, create a new Random Forest model which you will train on all training data
rf_model_on_full_data = RandomForestRegressor(random_state=1)


# fit rf_model_on_full_data on all data from the training data
rf_model_on_full_data.fit(X, y)
```

3. 导入需要预测的数据集

```python
# path to file you will use for predictions
test_data_path = '____'

# read test data file using pandas
test_data = pd.read_csv(test_data_path)

# create test_X which comes from test_data but includes only the columns you used for prediction.
# The list of columns is stored in a variable called features
test_X = test_data[melbourne_features]

# make predictions which we will submit.
test_preds = rf_model_on_full_data.predict(test_X)
```

4. 保存预测成果`ID`列及房价`SalePrice`列。另存为`csv`文件在根目录下

```python
# Run the code to save predictions in the format used for competition scoring

output = pd.DataFrame({'Id': test_data.Id,
                       'SalePrice': test_preds})
output.to_csv('submission.csv', index=False)
```
